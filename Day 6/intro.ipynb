```python
# Different ways to create an array using built in options

zeros_array=np.zeros(10,dtype=int)
print(zeros_array)
```

```python
# 2D Array

array_2=np.zeros((3,4),dtype=int)
print(array_2)
print(array_2.ndim)
print(array_2.shape)
print(array_2.dtype)
print(array_2.size)
```
