Encapsulation - It is data hiding.

Access modifiers

```python
class Sample:
    a=10 # public
    _b=20 # protected
    __c=30 # private

s1=Sample()
print(s1.a)
print(s1._b)
# print(s1.__c) can't be accessed.+
```

```python
class Sample:

    def m1(self):
        print("Public")
    def _m2(self):
        print("Protected")
    def __m3(self):
        print("Private.")
s1=Sample()
s1.m1()
s1._m2()
# s1.__m3() can't be accessed
```

```python
class Operation:
    def add(self,a,b):
        print("Addition of 2 numbers :",a+b)
    def add(self,a,b,c):
        print("Addition of 3 numbers : ",a+b+c)
    def add(self):
        print("Addition of 0 numbers")

op=Operation()
op.add()
```

```python
class Operation:
    def add(self,*args,**kwargs):
        print('Args',args)
        print('Kwargs',kwargs.values())
        print('--'*30)


op=Operation()
op.add(2,3,4,5)
op.add()
op.add(2,3)
op.add(a=100,b=200,c=300)
```

```python
try:
    print(2/1)
except ZeroDivisionError as msg:
    print("Error :",msg)
else:
    print("No Error")
finally:
    print("Execution complete")

```

```python
my_list=[i*i for i in range(1,21) if i%2==0]
print(my_list)

new_list=[num for num in range(100) if num%2==0]
print(new_list)
```

```python
div_by_3=list(map(lambda num: num*num if num>100 else num,new_list))
print(div_by_3)
```

```python
try:
    balance=10000
    withdraw=int(input("Enter amount to withdraw"))
    if withdraw>balance:
        raise ValueError("Insufficient balance")
except ValueError as msg:
    print("Transaction failed",msg)
else:
    balance-=withdraw
    print("Transaction successful. Remaining balance :",balance)
finally:
    print("Thank you for using ATM.")

```
