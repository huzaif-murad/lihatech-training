```python
import matplotlib.pyplot as plt
import numpy as np

X=np.array([2000,2005,2010,2015,2025])
y=np.array([3000,4000,3000,5000,6000])

X1=np.array([2000,2005,2010,2015,2025])
y1=np.array([2000,5000,4500,3000,4000])
plot_styles=dict(
     marker=".",
            markerfacecolor='red',
            mec="red",
            markersize=15,
            linestyle='solid',
            linewidth=3
         
)

plt.plot(X,y,**plot_styles)
plt.plot(X1,y1,**plot_styles)
plt.plot(X,y,**plot_styles)


```
