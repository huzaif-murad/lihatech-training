```python
import seaborn as sns

t=sns.load_dataset('titanic')
t[:2]
```

```python
#do a count plot to check 0 and 1 in survived
import matplotlib.pyplot as plt
# sns.countplot(x='survived',data=t)
sns.countplot(x='survived',data=t,hue='class')
```

```python
# histogram

sns.histplot(x='age',data=t,bins=[0,5,20,40,60,70,90,110],hue='class',palette='rainbow')
```

```python
sns.scatterplot(x='age',y='fare',data=t,hue='class',palette='Reds')
```

```python
sns.jointplot(x='age',y='fare',data=t,hue='class',palette='Reds')
```

```python
sns.jointplot(x='age',y='fare',data=t,palette='Reds',kind='hex')
```
