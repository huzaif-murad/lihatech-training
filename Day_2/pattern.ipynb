```python
for i in range(1,6):
    for j in range(1,i+1):
        print("*",end="")
    print("")
```

```python
for i in range(5,0,-1):
    for j in range(i,0,-1):
        print("*",end="")
    print("")
```

```python
n=5
for i in range(0,n):
    for j in range(0,n-i):
        print(" ",end="")
    for j in range(0,i*2-1):
        print("*",end="")
    print()
```

```python
for i in range(0,n):
    for j in range(0,n-i):
        print(" ",end="")
    for j in range(0,i):
        print("*",end="")
    print()
```
