```python
num_1=0
num_2=1
print(num_1,end=" ")
print(num_2,end=" ")
num=10
n=2
while(n<num):
    temp=num_1+num_2
    print(temp,end=" ")
    num_1=num_2
    num_2=temp
    n+=1
    
```

```python
num=int(input())

digit=0

while(num>0):
    digit+=1
    num=num//10

print(digit)
```

#### reverse a number using while loop

```python
num=int(input())
reverse=0
while(num>0):
    digit=num%10
    num=num//10
    reverse=(reverse*10) + digit
print(reverse)
```

#### sum of digits in a number

```python
num=int(input())
sum_of_digits=0

while(num>0):
    digit=num%10
    num=num//10
    sum_of_digits+=digit
print(sum_of_digits)
```
