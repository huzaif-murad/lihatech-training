```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
data=pd.read_csv("Salary_Data.csv")
data.head()
```

```python
import seaborn as sns

sns.scatterplot(data=data,x="YearsExperience",y='Salary',marker="*",color='red')
```

```python
#preapre data
#Seperate input and output

X=data[["YearsExperience"]]
y=data["Salary"]
```

```python
data.shape
```

#### For ML we use scikit learn library

```python
from sklearn.linear_model import LinearRegression

#to get access to the members of a class we need to create its object

model=LinearRegression()

model
```

```python
# Train the model -> fit() method

model.fit(X,y)
```

```python
model.coef_ # slope
```

```python
model.intercept_
```

```python


model.predict([[.6],[4],[0],[10]])
```

```python
model.predict([[5]])
```

```python

# y= mx + c

model.coef_ * 0 + model.intercept_
```

```python
test_data=[[0.2],[3],[6],[10],[12]]

y_pred=model.predict(test_data)
y_pred
```

```python
import matplotlib.pyplot as plt

plt.plot(test_data,y_pred,color='red',label='Prediction')
plt.plot(X,y,color="m",label="Actual Data")
plt.legend()
plt.show()
```
