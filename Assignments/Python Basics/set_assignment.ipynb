16. Write a Python program to find common unique words from two sentences using sets.

```python
sentence_1= "python is easy and powerful"
sentence_2= "python is popular and useful"

set_1=set(list(sentence_1.split()))
set_2=set(list(sentence_2.split()))

print(set_1.intersection(set_2))
```
