Write a Python program to print all alphabets from a to z. - using while loop

```python
ch='a'

while ch<='z':
    print(ch)
    ch=chr(ord(ch)+1)
```

Write a Python program to find sum of all even numbers between 1 to n.

```python
# nums=[num for num in range(1,101)]
# print(sum(nums))

sum=0

for i in range(1,101):
    sum+=i

print(sum)
```

Write a Python program to find sum of all odd numbers between 1 to n.

```python
# nums=[num for num in range(1,101,2)]
# # print(nums)
sum=0
for i in range(1,101,2):
    sum+=i
print(sum)
```

Write a Python program to count number of digits in a number.

```python
num=int(input("Enter the number to count the digits in it :"))
digits=0

while(num>0):
    digits+=1
    num=num//10

print(digits)
```
