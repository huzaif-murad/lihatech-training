5. Write a Python program to count the number of strings where the string length
is 2 or more and the first and last character are same from a given list of strings.
Sample List : ['abc', 'xyz', 'aba', '1221']

```python
my_list=['abc', 'xyz', 'aba', '1221']
count=0

for string in my_list:
    if len(string)>=2:
        if(string[0]==string[len(string)-1]):
            count+=1
print(count)

# new_list=list(filter(lambda string : string[0]==string[len(string)-1],filter(lambda string : len(string)>=2 ,my_list)))
# print(len(new_list))
```

7. Write a Python program to remove duplicates from a list.

```python
my_list=[2,3,4,3,2,1]
new_list=[]

for num in my_list:
    if num not in new_list:
        new_list.append(num)

print(new_list)
```

```python
def check_common(list1,list2):
    for l_1_member in list1:
        for l_2_member in list2:
            if(l_2_member==l_1_member):
                return True
    return False

list1=["hello","world",1,3,4]
list2=["orange","banana"]
ans=check_common(list1,list2)
print(ans)
```
