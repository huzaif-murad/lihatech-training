#### 03.Write a Python program to count the number of vowels and consonants in a given string.

```python
my_string="my name is Aman"
vowels_count=0
consonent_count=0
for i in my_string:
    if i.lower() in "aeiou":
        vowels_count+=1
    elif i!=" ":
        consonent_count+=1

print("Vowels count : ",vowels_count)
print("Consonent count : ",consonent_count)
```

#### 04.Write a Python program to count uppercase letters, lowercase letters, digits, and special characters in a string.

```python
sentence="Hello @123 World!!!"
upper_char=0
lower_char=0
digits=0
special_char=0

for letter in sentence:
    if letter.islower():
        lower_char+=1
    elif letter.isupper():
        upper_char+=1
    elif letter.isdigit():
        digits+=1
    elif letter.isspace():
        continue
    else:
        special_char+=1

print("Upper case characters :",upper_char)
print("Lower case characters :",lower_char)
print("Digits :",digits)
print("Special characters :",special_char)
```

#### 05.Write a Python program to check whether a given string is a palindrome.

```python
my_string="racecare"

# reverse_string=my_string[::-1]
# # print(reverse_string)

# if my_string == reverse_string:
#     print("Given string is a palindrome.")
# else:
#     print("Given string is not a palindrome.")

#*****************using two pointers*****************

left=0
right=len(my_string)-1
is_palindrome=True
while left<right:
    if my_string[left]!=my_string[right]:
        is_palindrome=False
        break
    left+=1
    right-=1

if is_palindrome:
    print("Given string is a palindrome.")
else:
    print("Given string is not a palindrome.")
```

#### 12.Write a Python program to find the most frequently occurring character in a string.

```python
my_string="his name is huzaifffff"

candidate=""
count=0

for char in my_string:
    if count==0:
        candidate=char
        count=1
    elif char==candidate:
        count+=1
    else:
        count-=1
print(candidate)
```

#### 16.Write a Python program to capitalize the first letter of every word in a sentence without using title().

```python
my_string="delhi is the capital of india"

is_starting=True
title_string=""
for i in range(len(my_string)):
    if my_string[i]==" ":
        is_starting=True
        title_string=title_string+" "
    elif is_starting:
        title_string=title_string+my_string[i].upper()
        is_starting=False
    else:
        title_string+=my_string[i]
print(title_string)
    
```
