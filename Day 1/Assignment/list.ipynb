#### 04.Write a Python program to count even and odd numbers in a list.

```python
my_list=[23,34,45,88,113,56,76]
even_count=0
odd_count=0

for num in my_list:
    if num%2==0:
        even_count+=1
    else:
        odd_count+=1

print("Even number count =",even_count)
print("Odd number count=",odd_count)
```

#### 06.Write a Python program to reverse a list without using the reverse() method.

```python
my_list=[23,34,45,56,75]

left=0
right=len(my_list)-1

while(left<right):
    temp=my_list[left]
    my_list[left]=my_list[right]
    my_list[right]=temp
    left+=1
    right-=1
print(my_list)
```

#### 08.Write a Python program to find the second largest number in a list.

```python
my_list=[32,221,12,21,56,54,55,100,102,222]

largest=my_list[0]
second_largest=my_list[0]

for num in my_list:
    if num>largest:
        second_largest=largest
        largest=num
    elif num>second_largest:
        second_largest=num
        

print("Second largest = ",second_largest)
```
