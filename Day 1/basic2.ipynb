### **Flow control blocks**

```python
# Selective or conditional execution/statements: if, elif, else

num = int(input("Enter a number: "))
if num > 0:
    print("The number is positive.")
elif num < 0:
    print("The number is negative.")
else:
    print("The number is zero.")
```

```python
age = int(input("Enter your age: "))
nationality = input("Enter your nationality: ")
nationality =nationality.lower()

if (age >= 18) and (nationality == "indian"):
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")
```

```python
nums = [1, 20, 3, 49, 50,9,12,43]
even = []
odd = []

for val in nums:
    if val%2==0: # check current value is even??
        even.append(val)
    else:
        odd.append(val)
print('Even list:',even)
print('Odd list:',odd) 
```

```python
marks = 'parth got 80 90 77 89 marks'
# total, % print
marks.split()
```

```python
'parth'.isdigit()
```

```python
final = []
for val in marks.split():
    if val.isdigit():
        final.append(int(val))
print('Total-',sum(final))
print('Percentage-',sum(final)/len(final),'%')
```
