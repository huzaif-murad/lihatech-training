📌 Problem 1: Remove Duplicates from a List

Given a list of integers, create a new list containing only unique elements while preserving their original order.

Input:

numbers = [4, 2, 4, 7, 2, 9, 7, 1]

Expected Output:

[4, 2, 7, 9, 1]

Requirements:

- Do not modify the original list.
- Do not use a set to remove duplicates.
- Handle empty lists and lists containing only duplicate values.
- Bonus: Analyze the time and space complexity of your solution.

```python
numbers = [4, 2, 4, 7, 2, 9, 7, 1]
new_list=[]
seen=set()

for num in numbers:
    if num not in seen:
        new_list.append(num)
        seen.add(num)
new_list
# for num in numbers:
#     if num not in new_list:
#         new_list.append(num)

# new_list
```

📌 Problem 2: Word Frequency Counter

Write a program to count how many times each word appears in a sentence, ignoring capitalization.

Input:

sentence = "Python is great and Python is powerful"

Expected Output:

{
    "python": 2,
    "is": 2,
    "great": 1,
    "and": 1,
    "powerful": 1
}

Requirements:

- Use a dictionary to store word frequencies.
- Handle extra spaces between words.
- Identify the most frequently occurring word.

```python
sentence = "Python is great and Python is powerful"

words=sentence.split()
frequency={}

for word in words:
    if word in frequency:
        frequency[word]+=1
    else:
        frequency[word]=1
        

print(frequency)
```

📌 Problem 4: Group Records by Category

Given a list of employee records, group employee names by department using a dictionary.

Input:

employees = [
    {"name": "Amit", "dept": "AI"},
    {"name": "Priya", "dept": "Data"},
    {"name": "Rahul", "dept": "AI"},
    {"name": "Neha", "dept": "Data"},
    {"name": "John", "dept": "Cloud"}
]

Expected Output:

{
    "AI": ["Amit", "Rahul"],
    "Data": ["Priya", "Neha"],
    "Cloud": ["John"]
}

Requirements:

- Use a loop and dictionary.
- Support any number of employees and departments.
- Preserve employee order within each department.
- Bonus: Explain how this pattern can be applied to grouping ML experiment results.

```python
employees = [
    {"name": "Amit", "dept": "AI"},
    {"name": "Priya", "dept": "Data"},
    {"name": "Rahul", "dept": "AI"},
    {"name": "Neha", "dept": "Data"},
    {"name": "John", "dept": "Cloud"}
]
group_by_dept={}
for record in employees:
    dept=record["dept"]

    if dept not in group_by_dept:
        group_by_dept[dept]=[]
    
    group_by_dept[dept].append(record["name"])

print(group_by_dept)
```

📌 Problem 5: Text Cleaning for NLP

Write a reusable function to preprocess text for a Natural Language Processing (NLP) pipeline.

Input:

text = "Python, Python! AI is GREAT."

Expected Output:

["python", "python", "ai", "is", "great"]

Requirements:

- Convert text to lowercase.
- Remove punctuation without joining separate words.
- Split the cleaned text into individual words.
- Handle empty strings and whitespace-only input.
- Bonus: Consider how you would handle Unicode text in a production NLP pipeline.

```python
text = "Python, Python! AI is GREAT."

def text_cleaning(text):
    text=text.lower()
    print(text)
    cleaned=" "
    for ch in text:
        if(ch.isalnum() or ch.isspace()):
            cleaned+=ch
        else:
            cleaned+=" "
    return cleaned.split()

print(text_cleaning(text))
    
```
