```python
#Order matters in positional arguments if we change the order then output is unexpected, also number of the arguments also must be same as in function defination
def display(name,age,salary):
    print("name",name)
    print("age",age)
    print("salary",salary)

display("Huzaif",23,20000)
```

```python
#solution to above problem is keyword argument
# Key word argument

#key value association


display(name="Huzaif",salary=3000,age=23)
```

```python
#Default Arguments
def account(acc_nm='guest',branch="SBI"):
    print("Welcome",acc_nm,"to",branch)

account("Huzaif","ICICI")
```

```python
#variable length arguments

# *args - var len positional arguments
# **kwrgs - var len keyword arguments

def test(*num):
    print(num)

test()
test(23,34)
test(1)
test(1,2,3,4)

```

#### A function who need another function as an input called as higher order function
1. map(function,iterable)
2. filter(function,iterable)
3. reduce(function,iterable)

```python
data=['male','female','male','female','male','female','male','female']
new_list=list(map(lambda gender:0 if gender=="male" else 1,data))
print(new_list)
```

```python
nums=[2,3,5,6,19,10,22,50,33]

square=list(map(lambda num: num*num if num>10 else num ,nums))
print(square)
```

```python
nums=[10,20,30,40,50]
new_nums=list(filter(lambda num: num>25,nums))
print(new_nums)
```

```python
nums=[]
for i in range(1,31):
    nums.append(i)
# print(nums)
new_nums=list(filter(lambda num: num%7==0 or num%5==0,nums))
print(new_nums)
```

#### Reduce

```python
from functools import reduce

nums=[1,2,3,4,5]

num=reduce(lambda a,b:a+b,nums)
print(num)
maximum=reduce(lambda a,b:max(a,b),nums)
print(maximum)
```
